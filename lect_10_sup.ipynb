$$
\frac{\partial y } {\partial x} = f(x,y) = sin(x) \cdot cos(y)
$$

$$
y_{i+1} = y_i +\frac{\partial y } {\partial x}\Delta x =  y_i + f(x,y) \Delta x
$$

```python
import plotly.graph_objects as go
import numpy as np
import pandas as pd

point_x = 1
point_y = -2
point_z = np.sin(point_x) * np.cos(point_y)+0.05
print(f"point_x     = {point_x}")
print(f"point_y     = {point_y}")
print(f"point_z     = {point_z-0.05}")

point = go.Scatter3d(
    x=[point_x],
    y=[point_y],
    z=[point_z],
    mode='markers',
    marker=dict(size=5, color='lightgreen'),
    name='Movable Point'
)

x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
x_data, y_data = np.meshgrid(x, y)

# Define the function
z_data = np.sin(x_data) * np.cos(y_data)
surf = go.Surface(x=x_data, y=y_data, z=z_data)
fig = go.Figure(data=[surf,point])

fig.update_layout(
    title=dict(
        text='sin(x) * cos(y)',), 
        autosize=False,
        width=1000, 
        height=1000, 
        margin=dict(l=65, r=50, b=65, t=90), 
    scene=dict(
        xaxis_title='X Axis',
        yaxis_title='Y Axis',
        zaxis_title='Z Axis',
        aspectmode='manual',  # 비율을 수동으로 설정
        aspectratio=dict(x=1, y=1, z=0.2),  # x:y:z 비율 설정
        # scaleanchor='x' 
    )
)  

fig.show()
```

$$
x_i = 1, y_i = -2
$$

$$
\frac{\partial y } {\partial x} = f(1,-2) = sin(1) \cdot cos(-2) = -0.35
$$

$$
\Delta x = 0.5
$$

$$
y_{i+1} = y_i +\frac {\partial y} {\partial x}\Delta x = -2 + f(1,-2) \cdot\Delta x = -2 -0.35 \cdot 0.5 = -2.17508774
$$
 
 Thus, $$x_{i+1} = 1.5, y_{i+1}^0 = -2.17508774$$ 

$$
F(x,y) = \log | \sec(y) + \tan (y) | + \cos(x) - C
$$

$$
C = \log | \sec(-2) + \tan (-2) | + \cos(1)
$$

```python
import math
def sec(x):
    if x % (math.pi / 2) == 0:
        raise ValueError("sec(x) is undefined for x = {}.".format(x))
    return 1 / math.cos(x)

C = math.log(abs(sec(-2) + math.tan(-2))) + math.cos(1)
print("C = ", C)

def F(x, y):
    return math.log(abs(sec(y) + math.tan(y))) + math.cos(x) - C

print(f"F(1, -2) = {F(1, -2)}")
print(f"F(1.5, -2.17508774) = {F(1.5, -2.17508774)} != 0")
print(f"F(1.5, -2.2416) = {F(1.5, -2.2416):0.8f} =~ 0")

# for i in range(1, 10):
#     x = 1.5
#     y = -1.99127911
#     y_new = y - F(x, y) / (F(x, y + 0.0001) - F(x, y)) * 0.0001
#     y_new = -2.241 - 0.0001 * i
#     print(f"Iteration {i}: x = {x}, y_new = {y_new}, F(x, y_new) = {F(x, y_new)}")
#     if abs(F(x, y_new)) < 1e-6:
#         break

```
